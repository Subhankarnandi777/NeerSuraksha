```python
def find_runs(df, station, value):
    temp = df[df["Station"] == station].copy()
    temp = temp.sort_values("Timestamp").reset_index(drop=True)

    temp["is_target"] = temp["Groundwater_Level"].eq(value)

    # Start a new run whenever target status changes
    temp["run_id"] = temp["is_target"].ne(
        temp["is_target"].shift()
    ).cumsum()

    runs = (
        temp[temp["is_target"]]
        .groupby("run_id")
        .agg(
            Start=("Timestamp", "min"),
            End=("Timestamp", "max"),
            Count=("Timestamp", "size")
        )
        .sort_values("Count", ascending=False)
    )

    return runs


for station, value in suspicious.items():
    runs = find_runs(clean_df, station, value)

    print("\n" + "="*70)
    print(station, "| Value:", value)
    print("Longest repeated runs:")

    print(runs.head(10))
```

```python
# Sort properly
clean_df = clean_df.sort_values(
    ["Station", "Timestamp"]
).reset_index(drop=True)

# Previous value
clean_df["Previous_Level"] = (
    clean_df.groupby("Station")["Groundwater_Level"].shift(1)
)

# Same as previous reading
clean_df["Same_As_Previous"] = (
    clean_df["Groundwater_Level"] ==
    clean_df["Previous_Level"]
)

# Create run ID for consecutive identical values
clean_df["Value_Change"] = (
    clean_df["Groundwater_Level"] !=
    clean_df["Previous_Level"]
)

clean_df["Run_ID"] = (
    clean_df.groupby("Station")["Value_Change"]
    .cumsum()
)

# Size of each repeated-value run
clean_df["Run_Length"] = (
    clean_df.groupby(["Station", "Run_ID"])["Groundwater_Level"]
    .transform("size")
)

print(clean_df["Run_Length"].describe())
```

```python
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# STEP 1: CLASSIFY MISSING RUNS
# ------------------------------------------------------------

df = clean_df.copy()

df["Timestamp"] = pd.to_datetime(df["Timestamp"])
df = df.sort_values(["Station", "Timestamp"]).reset_index(drop=True)

# Missing = no final usable groundwater value
df["Is_Missing"] = df["Groundwater_Level_Final"].isna()

# Create a new run ID whenever missing/non-missing status changes
df["Missing_Run_ID"] = (
    df.groupby("Station")["Is_Missing"]
      .transform(lambda x: x.ne(x.shift()).cumsum())
)

# Keep only missing observations
missing_df = df[df["Is_Missing"]].copy()

# Calculate size of each missing run
run_sizes = (
    missing_df
    .groupby(["Station", "Missing_Run_ID"])
    .size()
    .rename("Missing_Readings")
    .reset_index()
)

# Add run size back to each missing observation
df = df.merge(
    run_sizes,
    on=["Station", "Missing_Run_ID"],
    how="left"
)

# ------------------------------------------------------------
# CLASSIFY GAP SIZE
# ------------------------------------------------------------

def classify_gap(n):
    if pd.isna(n):
        return "NO_GAP"
    elif n <= 4:
        return "SHORT_GAP"
    elif n <= 20:
        return "MEDIUM_GAP"
    else:
        return "LONG_GAP"

df["Gap_Class"] = df["Missing_Readings"].apply(classify_gap)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

gap_summary = (
    df[df["Is_Missing"]]
    .groupby(["Station", "Gap_Class"])
    .size()
    .unstack(fill_value=0)
)

print("=" * 70)
print("MISSING GAP CLASSIFICATION")
print("=" * 70)

display(gap_summary)

print("\nOverall gap-class distribution:")
print(
    df.loc[df["Is_Missing"], "Gap_Class"]
      .value_counts()
)

print("\nTotal missing observations:")
print(df["Is_Missing"].sum())
```

```python
# ------------------------------------------------------------
# GAP-LEVEL SUMMARY
# ------------------------------------------------------------

gap_table = (
    df[df["Is_Missing"]]
    .groupby(["Station", "Missing_Run_ID"])
    .agg(
        Start=("Timestamp", "min"),
        End=("Timestamp", "max"),
        Missing_Readings=("Timestamp", "size"),
        Duration_Hours=(
            "Timestamp",
            lambda x: (x.max() - x.min()).total_seconds() / 3600
        )
    )
    .reset_index()
)

gap_table["Gap_Class"] = gap_table["Missing_Readings"].apply(classify_gap)

print("Number of missing runs:", len(gap_table))

print("\nGap class counts:")
print(gap_table["Gap_Class"].value_counts())

print("\nLargest gaps:")
display(
    gap_table
    .sort_values("Missing_Readings", ascending=False)
    .head(30)
)
```

```python
# ============================================================
# STATION-WISE FINAL DATA QUALITY / AVAILABILITY REPORT
# ============================================================

import pandas as pd
import numpy as np

df = clean_df.copy()

# Make sure timestamp is datetime
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Sort properly
df = df.sort_values(["Station", "Timestamp"]).reset_index(drop=True)

# ------------------------------------------------------------
# 1. Valid / missing counts
# ------------------------------------------------------------

availability = (
    df.groupby("Station")
      .agg(
          Total_Records=("Timestamp", "size"),
          Valid_Records=("Groundwater_Level_Final", "count"),
          Missing_Records=("Groundwater_Level_Final", lambda x: x.isna().sum()),
          Start_Date=("Timestamp", "min"),
          End_Date=("Timestamp", "max")
      )
)

availability["Valid_%"] = (
    availability["Valid_Records"]
    / availability["Total_Records"] * 100
)

availability["Missing_%"] = (
    availability["Missing_Records"]
    / availability["Total_Records"] * 100
)

# ------------------------------------------------------------
# 2. Number of gaps
# ------------------------------------------------------------

gap_counts = (
    df.assign(
        is_missing=df["Groundwater_Level_Final"].isna()
    )
    .groupby("Station")["is_missing"]
    .sum()
    .rename("Missing_Observations")
)

availability = availability.drop(
    columns=["Missing_Records"]
).join(gap_counts)

availability["Missing_Observations"] = (
    availability["Missing_Observations"].astype(int)
)

# ------------------------------------------------------------
# 3. Expected 6-hour interval check
# ------------------------------------------------------------

df["Time_Diff_Hours"] = (
    df.groupby("Station")["Timestamp"]
      .diff()
      .dt.total_seconds()
      / 3600
)

temporal = (
    df.groupby("Station")
      .agg(
          Median_Interval_Hours=("Time_Diff_Hours", "median"),
          Min_Interval_Hours=("Time_Diff_Hours", "min"),
          Max_Interval_Hours=("Time_Diff_Hours", "max"),
          Irregular_Intervals=(
              "Time_Diff_Hours",
              lambda x: ((x.notna()) & (x != 6)).sum()
          )
      )
)

# ------------------------------------------------------------
# 4. Longest missing run
# ------------------------------------------------------------

def longest_missing_run(group):

    missing = group["Groundwater_Level_Final"].isna()

    if not missing.any():
        return 0

    run_id = missing.ne(missing.shift()).cumsum()

    run_lengths = (
        missing.groupby(run_id)
        .sum()
    )

    return int(run_lengths.max())


longest_gap = (
    df.groupby("Station")
      .apply(longest_missing_run, include_groups=False)
      .rename("Longest_Missing_Run")
)

# ------------------------------------------------------------
# 5. Combine
# ------------------------------------------------------------

availability = (
    availability
    .join(temporal)
    .join(longest_gap)
)

# Sort by missing percentage
availability = availability.sort_values(
    "Missing_%",
    ascending=False
)

print("=" * 80)
print("FINAL STATION-WISE DATA QUALITY REPORT")
print("=" * 80)

display(availability.round(2))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# FINAL DATASET PREPARATION
# ============================================================

final_df = clean_df.copy()

# ------------------------------------------------------------
# 1. Standardize timestamp
# ------------------------------------------------------------
final_df["Timestamp"] = pd.to_datetime(
    final_df["Timestamp"],
    errors="coerce"
)

# ------------------------------------------------------------
# 2. Sort data
# ------------------------------------------------------------
final_df = final_df.sort_values(
    ["Station", "Timestamp"]
).reset_index(drop=True)

# ------------------------------------------------------------
# 3. Remove exact duplicate station-timestamps
# ------------------------------------------------------------
duplicate_count = final_df.duplicated(
    subset=["Station", "Timestamp"]
).sum()

print("Duplicate station-timestamps before removal:", duplicate_count)

final_df = final_df.drop_duplicates(
    subset=["Station", "Timestamp"],
    keep="first"
).reset_index(drop=True)

# ------------------------------------------------------------
# 4. Ensure final groundwater column is numeric
# ------------------------------------------------------------
final_df["Groundwater_Level_Final"] = pd.to_numeric(
    final_df["Groundwater_Level_Final"],
    errors="coerce"
)

# ------------------------------------------------------------
# 5. Ensure quality flag exists
# ------------------------------------------------------------
if "Quality_Flag" not in final_df.columns:
    final_df["Quality_Flag"] = np.where(
        final_df["Groundwater_Level_Final"].notna(),
        "OK",
        "GAP"
    )

# ------------------------------------------------------------
# 6. Any row without a final value must be GAP
# ------------------------------------------------------------
final_df.loc[
    final_df["Groundwater_Level_Final"].isna(),
    "Quality_Flag"
] = "GAP"

# ------------------------------------------------------------
# 7. Make sure valid final observations are OK
#    unless explicitly flagged otherwise
# ------------------------------------------------------------
final_df.loc[
    final_df["Groundwater_Level_Final"].notna()
    & final_df["Quality_Flag"].isna(),
    "Quality_Flag"
] = "OK"

# ------------------------------------------------------------
# 8. Final column order
# ------------------------------------------------------------
preferred_columns = [
    "Station",
    "Timestamp",
    "Groundwater_Level_Raw",
    "Groundwater_Level_Final",
    "Quality_Flag"
]

final_columns = [
    c for c in preferred_columns
    if c in final_df.columns
]

final_df = final_df[final_columns]

print("\nFinal dataset shape:", final_df.shape)
```
